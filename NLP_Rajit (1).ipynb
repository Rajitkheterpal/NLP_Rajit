```python
sentences = [
    "I like football",
    "The weather is very hot",
    "I love eating pizza",
    "My laptop is very fast",
    "I am going to college",
    "The movie was very interesting"
]
```

```python
labels = [
    "sports",
    "weather",
    "food",
    "technology",
    "education",
    "entertainment"
]
```

```python
# Tokenization using whitespace

tokenized_sentences = []

for sentence in sentences:
    tokens = sentence.split(" ")
    tokenized_sentences.append(tokens)

print(tokenized_sentences)
```

```python
words = []
for sentence in sentences:
    words.extend(sentence.split())
print(words)
```

```python
# Total number of words
total_words = len(words)

# Find unique words
vocabulary = set(words)

# Number of unique words
vocab_count = len(vocabulary)

print("Total word count:", total_words)
print("Vocabulary count:", vocab_count)
```

```python
print("Vocabulary:")
print(vocabulary)
```

```python
# Convert the vocabulary into a list
vocabulary = sorted(set(words))

# Assign a unique number to each word
word_to_id = {}

for i, word in enumerate(vocabulary):
    word_to_id[word] = i

print(word_to_id)
```

```python
# Convert the words in each sentence into numerical IDs
vectorized_sentences = []

for sentence in sentences:
    sentence_words = sentence.split()
    sentence_vector = [word_to_id[word] for word in sentence_words]
    vectorized_sentences.append(sentence_vector)

for vector in vectorized_sentences:
    print(vector)
```

```python

def one_hot_encode(word, vocabulary):
    vector = [0] * len(vocabulary)
    index = vocabulary.index(word)
    vector[index] = 1
    return vector
```

```python

print("Vocabulary size:", len(vocabulary))

print("One-hot vector for 'football':")
print(one_hot_encode("football", vocabulary))

print("Vector length:", len(one_hot_encode("football", vocabulary)))
```

```python

word_to_vector = {}

for word in vocabulary:
    word_to_vector[word] = one_hot_encode(word, vocabulary)

print(word_to_vector)
```

```python
def encode_sentences(sentences, word_to_vector):
    all_vectors = []

    for sentence in sentences:
        tokens = sentence.split(" ")
        sentence_vectors = []

        for word in tokens:
            sentence_vectors.append(word_to_vector[word])

        all_vectors.append(sentence_vectors)

    return all_vectors


# Convert all 6 sentences
encoded_sentences = encode_sentences(sentences, word_to_vector)

# Display each sentence and its vectors
for i, vectors in enumerate(encoded_sentences):
    print(f"Sentence {i + 1}: {sentences[i]}")
    print("Number of word vectors:", len(vectors))
    print("Vector size:", len(vectors[0]))
    print(vectors)
    print()
```

```python

import numpy as np

sentence_matrices = []

for sentence in sentences:
    tokens = sentence.split(" ")

    # Get the one-hot vector for every word
    vectors = [word_to_vector[word] for word in tokens]

    # Stack the vectors into a matrix
    matrix = np.array(vectors)

    sentence_matrices.append(matrix)

# Display each sentence matrix and its dimensions
for i, matrix in enumerate(sentence_matrices):
    print(f"\nSentence {i + 1}: {sentences[i]}")
    print(matrix)
    print("Matrix shape:", matrix.shape)
```
